```python
#Q1. Data Cleaning: Missing Values
#Loading the student-data-v2.csv file
import pandas as pd
df = pd.read_csv('student-data-v2.csv')
#Prints truncated version of the dataframe
print(df)
#Prints the Full Data
print(df.to_string())
#Total Missing Values in DataFrame
Total_Missing_Values = df.isnull().sum().sum()
print("Total_Missing_Values:" ,Total_Missing_Values)
#Filling the missing values of each column with the mean of that column
#Attendance Marks Column
df['attendance'] = df['attendance'].fillna(df['attendance'].mean())
print(df.to_string())
#Python Marks Column
df['python_marks'] = df['python_marks'].fillna(df['python_marks'].mean())
print(df.to_string())
```

```python
#Q2. Data Cleaning: Duplicates 
# df is created in the earlier loading cell; remove duplicate rows and keep a clean copy
df_cleaned = df.drop_duplicates().copy()

print(df_cleaned.head())

row_count = len(df_cleaned)

print(f"Number of unique rows: {row_count}")
```

```python
#Q3— Data Cleaning: Standardizing Text
#Removing leading and trailing spaces from the department column
df_cleaned['department'] = df_cleaned['department'].str.strip()
print(df_cleaned.to_string())
#Converting all department names to uppercase
df_cleaned['department'] = df_cleaned['department'].str.upper()
print(df_cleaned.to_string())
#Number of AIML students after cleaning the department column
AIML_students = df_cleaned[df_cleaned['department'] == 'AIML']
print(f"Number of AIML students: {len(AIML_students)}")
#printing the cleaned dataframe
print(df_cleaned.to_string())
```

```python
# Q4 — Merging DataFrames

# Load mentor feedback data
df_mentor = pd.read_csv('mentor-feedback.csv')

# Display mentor data
print(df_mentor.to_string())

# Merge student and mentor DataFrames
df_merged = pd.merge(
    df_cleaned,
    df_mentor,
    on='student_id'
)

print(df_merged.to_string())

# Left join: keep ALL students,
# even if they don't have mentor feedback
df_merged_left = pd.merge(
    df_cleaned,
    df_mentor,
    on='student_id',
    how='left'
)

print(df_merged_left.to_string())
#Number of students with missing mentor feedbackafter the left join
missing_feedback_students = df_merged_left[df_merged_left['feedback_score'].isnull()]
print(f"Number of students with missing mentor feedback: {len(missing_feedback_students)}")
```

```python
# Q5 — GroupBy on Merged Data
#calculating the avrage feedback score for each department
avg_feedback_by_department = df_merged_left.groupby('department')['feedback_score'].mean()
print(avg_feedback_by_department)
#Highest average feedback score among departments
highest_avg_feedback_department = avg_feedback_by_department.idxmax()
print(f"Department with highest average feedback score: {highest_avg_feedback_department}")
#Whole number of highest average feedback score
highest_avg_feedback_score = avg_feedback_by_department.max()
print(f"Highest average feedback score: {highest_avg_feedback_score}")
#Converting  highest average feedbak score to a whole number
highest_avg_feedback_score = int(highest_avg_feedback_score)
print(f"Highest average feedback score (as whole number): {highest_avg_feedback_score}")
```

```python
# Q6 — Dates

# Convert join_date to Pandas datetime format
df_merged_left['join_date'] = pd.to_datetime(
    df_merged_left['join_date'],
    dayfirst=True
)

# Define August 2023 date range
august_start = pd.Timestamp('2023-08-01')
august_end = pd.Timestamp('2023-08-31')

# Find students who joined during August 2023
august_joins = df_merged_left[
    (df_merged_left['join_date'] >= august_start) &
    (df_merged_left['join_date'] <= august_end)
]

# Count students
print(f"Number of students who joined in August 2023: {len(august_joins)}")
```
